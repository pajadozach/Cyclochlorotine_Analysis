# Ligand Docking

## APO Blind Docking

### Enniatin A

```python
%%bash
for i in $(seq -w 01 027); do
    vina --receptor HSA_Target/1AO6/1ao6.pdbqt \
         --ligand Ligand_Pubchem_Data/EnniatinA/EnniatinA.pdbqt \
         --config dockprep/box_${i}.txt \
         --exhaustiveness=8 --out Results/APO/ENA/ENA_${i}.pdbqt

    obabel Results/APO/ENA/ENA_${i}.pdbqt -O Results/APO/ENA/ENA_${i}_.pdb -m

done
```
