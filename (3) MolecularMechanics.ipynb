```python
tleap_template = """source leaprc.protein.ff14SB
source leaprc.gaff

rec = loadpdb {receptor_pdb}
lig = loadmol2 {ligand_mol2}
loadamberparams {ligand_frcmod}

complex = combine {{rec lig}}

saveamberparm rec {receptor_prmtop} {receptor_inpcrd}
saveamberparm lig {ligand_prmtop} {ligand_inpcrd}
saveamberparm complex {complex_prmtop} {complex_inpcrd}

quit
"""

mmpbsa_template = """&general
  startframe=1, endframe=1, interval=1,
  verbose=1,
/
&gb
  igb=5, saltcon=0.150
/
"""
```

```python
import subprocess, os
from pathlib import Path

# --- mapping of receptor -> ligand lists (use your variables above)
pairs = [
    # receptor p2bxd (use prepared file p2bxd)
    (p2bxd, [h1ea, h1ea1, h1eb, h1eb1, h1rwf, h1rbp]),
    # receptor p2bxg
    (p2bxg, [h2ea, h2ea1, h2eb, h2eb1, h2rwf, h2ibp]),
    # receptor p1ao6 (site1)
    (p1ao6, [a1ea, a1ea1, a1eb, a1eb1, a1rwf, a1ibp]),
    # receptor p1ao6 (site2)
    (p1ao6, [a2ea, a2ea1, a2eb, a2eb1, a2rwf, a2ibp]),
]


def safe_name(path):
    return Path(path).stem.replace('.', '_').replace('/', '_')

for receptor, lig_list in pairs:
    receptor_base = safe_name(receptor)
    for lig in lig_list:
        lig_base = safe_name(lig)
        case_name = f"{receptor_base}__{lig_base}"
        workdir = outdir / case_name
        workdir.mkdir(parents=True, exist_ok=True)
        print(f"\n=== Processing {case_name} ===")

        # 1) Step: parameterize ligand with antechamber
        mol2 = workdir / f"{lig_base}.mol2"
        frcmod = workdir / f"{lig_base}.frcmod"
        cmd_ac = [ANTECHAMBER, "-i", lig, "-fi", "mol2", "-o", str(mol2), "-fo", "mol2", "-c", "bcc", "-s", "2"]
        subprocess.run(cmd_ac, check=True)

        # parmchk2
        cmd_pc = [PARMCHK2, "-i", str(mol2), "-f", "mol2", "-o", str(frcmod)]
        subprocess.run(cmd_pc, check=True)

        # 2) Step: write tleap.in and run tleap to create prmtops
        receptor_prmtop = workdir / "receptor.prmtop"
        receptor_inpcrd = workdir / "receptor.inpcrd"
        ligand_prmtop = workdir / "ligand.prmtop"
        ligand_inpcrd = workdir / "ligand.inpcrd"
        complex_prmtop = workdir / "complex.prmtop"
        complex_inpcrd = workdir / "complex.inpcrd"

        tleap_in = workdir / "tleap.in"
        tleap_text = tleap_template.format(
            receptor_pdb=os.path.abspath(receptor),
            ligand_mol2=os.path.abspath(mol2),
            ligand_frcmod=os.path.abspath(frcmod),
            receptor_prmtop=os.path.abspath(receptor_prmtop),
            receptor_inpcrd=os.path.abspath(receptor_inpcrd),
            ligand_prmtop=os.path.abspath(ligand_prmtop),
            ligand_inpcrd=os.path.abspath(ligand_inpcrd),
            complex_prmtop=os.path.abspath(complex_prmtop),
            complex_inpcrd=os.path.abspath(complex_inpcrd),
        )
        tleap_in.write_text(tleap_text)

        subprocess.run([TLEAP, "-f", str(tleap_in)], check=True)

        # 3) Step: write mmpbsa.in (single frame) and run MMPBSA.py
        mmpbsa_in = workdir / "mmpbsa.in"
        mmpbsa_in.write_text(mmpbsa_template)

        out_file = workdir / f"FINAL_RESULTS_{case_name}.dat"
        cmd_mmpbsa = [
            MMPBSA, "-O",
            "-i", str(mmpbsa_in),
            "-o", str(out_file),
            "-cp", str(complex_prmtop),
            "-rp", str(receptor_prmtop),
            "-lp", str(ligand_prmtop)
            # no -y (no trajectory) -> single-frame
        ]
        print("Running MMPBSA.py (single frame) ...")
        subprocess.run(cmd_mmpbsa, check=True)

        print(f"Done: results -> {out_file}")
```
