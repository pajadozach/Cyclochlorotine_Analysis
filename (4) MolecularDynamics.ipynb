```python
gmx grompp -f minim_ulit.mdp -c em.gro -r em.gro -p MD/protein/p2bxdtopol_h1ea.top -o emulit.tpr
gmx mdrun -deffnm emulit -ntmpi 1 -ntomp 6 -pin on -v

'Sanity check if lig is in the right location'

'make Protein_LIG entry in ndx'

gmx make_ndx -f emulit.tpr -o init_index.ndx

gmx distance -s init.tpr -f MD/box_2bxd_h1ea.gro -n init_index.ndx \
  -select 'com of group "Protein" plus com of group "UNL"' -oall init_com.xvg
cat init_com.xvg

'Terminal'
gmx grompp -f nvt.mdp -c emulit.gro -r emulit.gro -p MD/protein/p2bxdtopol_h1ea.top -o nvt.tpr
gmx mdrun -deffnm nvt -ntmpi 1 -ntomp 6 -pin on -v

gmx grompp -f npt.mdp -c nvt.gro -r nvt.gro -p MD/protein/p2bxdtopol_h1ea.top -o npt.tpr
gmx mdrun -deffnm npt -ntmpi 1 -ntomp 6 -pin on -v

gmx grompp -f md.mdp -c npt.gro -p MD/protein/p2bxdtopol_h1ea.top -o md.tpr
gmx mdrun -deffnm md -ntmpi 1 -ntomp 6 -pin on -v

gmx grompp -f energy_md.mdp -c md.gro -p MD/protein/p2bxdtopol_h1ea.top -o energy.tpr
gmx mdrun -s energy.tpr -rerun md.xtc -deffnm energy -ntmpi 1 -ntomp 6 -pin on -v

gmx energy -f energy.edr -o MD/ENA/pl_interaction.xvg


gmx make_ndx -f md.tpr -o MD/ENA/index.ndx


gmx trjconv -s md.tpr -f md.xtc -o md_nojump.xtc -pbc nojump 'Sytem Output'

gmx trjconv -s md.tpr -f md_nojump.xtc -o md_whole.xtc -pbc whole 'Sytem Output'

gmx trjconv -s md.tpr -f md_whole.xtc -o md_center.xtc -center -pbc mol 'Center protein; System Output'

gmx trjconv -s md.tpr -f md_center.xtc -o md_fit.xtc -fit rot+trans 'Fit protein; System Output'

'back to root' 'activate gmxMMPBSA' 
conda activate gmxMMPBSA

'back to working directory'


mpirun -np 8 gmx_MMPBSA -O -i mmpbsa.in -cs energy.tpr -ci MD/ENA/index.ndx -cg 1 13 -ct md_fit.xtc -cp MD/protein/p2bxdtopol_h1ea.top -o MD/ENA/RESULTS_MMPBSA_recalc.dat -nogui
```
